````python
from dataclasses import dataclass, field
from enum import Enum

from array import array

import numpy as np
import pandas as pd
from rich.table import Table

import rich_struct as rs
from rich_struct import print_md, print_json, print_xml, print_obj, print_yaml, print_syntax, print_table
from rich_struct import print_html, print_list
from rich_struct import ThemeName

MARKDOWN = """# Project Documentation

Welcome to the **project documentation**.

This theme uses H1 as the brightest and most prominent heading,
with H2 slightly calmer and H3 more subtle.

## Getting Started

Follow the steps below to set up your environment.

### Prerequisites

- Install **Rust** and `cargo`
- Have Docker installed
- Basic knowledge of *Linux*
- Optional: a GitHub account

### Installation

1. Clone the repository
2. Install dependencies
3. Configure the application
4. Start the service

```bash
git clone https://example.com/project.git
cd project
cargo build
cargo run
```

### Next Steps

Continue with the [installation guide](https://example.com/docs/install).

> 💡 **Tip:** Keep your configuration in environment variables instead of
> committing secrets directly to the repository.

---

## Configuration

### Rust Example

```rust
fn main() {
    println!("Hello, world!");

    // This is a comment
    let numbers = vec![1, 2, 3];

    for number in numbers {
        println!("Number: {}", number);
    }
}
```

### Table Example

| Feature | Description | Status |
|---|---|---|
| Fast | High performance | ✅ Stable |
| Safe | Memory safety | ✅ Stable |
| Async | Tokio-based asynchronous runtime | 🚀 Active |
| Cross-platform | Linux, macOS and Windows | ✅ Stable |
| Plugins | Optional extension system | 🧪 Experimental |

---

## Notes

You can combine **bold text**, *italic text*, ~~strikethrough~~
and `inline code` inside normal paragraphs.

Useful commands:

- `cargo build`
- `cargo test`
- `cargo run`
- `docker compose up`

> **Warning:** Never commit passwords, API keys, or production credentials
> directly into configuration files.

### Final Check

- [x] Build succeeds
- [x] Tests pass
- [x] Configuration is valid
- [ ] Production deployment completed
"""

JSON = {"name": "rich-struct", "version": 0.1, "tags": ["print", "theme"], "stable": False, "license": None}

YAML = """# deployment
service:
  name: api
  replicas: 3
  debug: false
  ports: [80, 443]
  image: "ghcr.io/acme/api:1.2"
"""

HTML = """<html>
<head><title>Release notes</title><style>p { color: red }</style></head>
<body>
<h1>Release notes</h1>
<p>Version <b>0.4.0</b> adds <i>two</i> helpers, <code>print_html</code> and <code>print_list</code>.
See the <a href="https://example.com/docs">documentation</a>.</p>
<h2>Changes</h2>
<ul><li>HTML is rendered<ul><li>its tags are not shown</li></ul></li><li>CSS and scripts are dropped</li></ul>
<ol><li>Install the <code>html</code> extra</li><li>Call <code>print_html</code></li></ol>
<blockquote>Rendered through Markdown, so it follows the theme.</blockquote>
<pre><code class="language-python">def greet(name):
    return f"Hello, {name}!"  # highlighted as Python</code></pre>
<table>
<tr><th>helper</th><th>input</th></tr>
<tr><td>print_html</td><td>HTML</td></tr>
<tr><td>print_list</td><td>sequences</td></tr>
</table>
<script>alert("never shown")</script>
</body>
</html>
"""

XML = """<?xml version="1.0"?>
<!-- catalogue -->
<catalog updated="2026-10-11">
  <book id="bk101" lang="en">
    <title>Printing structs</title>
    <price>44.95</price>
  </book>
</catalog>
"""


class Role(Enum):
    ADMIN = 1
    GUEST = 2


@dataclass
class Address:
    city: str
    postcode: str


@dataclass
class User:
    name: str
    age: int
    active: bool = True
    email: str | None = None
    role: Role = Role.ADMIN
    address: Address | None = None
    tags: list[str] = field(default_factory=list)
    settings: dict = field(default_factory=dict)


OBJECT = User(
    "Ada",
    36,
    address=Address("Kraków", "30-001"),
    tags=["admin", "dev"],
    settings={"theme": "dark", "retries": 3, "ratio": 0.75},
)

TABLE = Table("theme", "kind", title="A rich renderable")
TABLE.add_row("Tokyo Night", "dark")
TABLE.add_row("Light+", "light")

RECORDS = [
    {"id": 1, "name": "Ada", "score": 91.5, "admin": True},
    {"id": 2, "name": "Linus", "score": 78.25, "admin": False},
    {"id": 3, "name": "Grace", "score": None, "admin": False},
]

ARRAY = np.random.default_rng(0).normal(size=(4, 5))

DF = pd.DataFrame(
    {
        "city": ["Kraków", "Oslo", "Lima"],
        "population": [804_237, 709_037, 10_092_000],
        "area_km2": [326.8, 454.0, 2672.3],
        "capital": [False, True, True],
        "twinned": pd.to_datetime(["1991-05-18", "2003-09-01", None]),
    }
).set_index("city")

SERIES = DF["area_km2"]

SEQUENCE = [123, 1444.555, "Test text", 4345.666, None, True, *np.random.default_rng(0).normal(size=200).round(3)]

````

## Table inputs

`print_table` takes lists, arrays, numpy arrays, pandas DataFrames and Series. Every input shape once, in the default theme; the per-theme comparisons follow.

````python
print_table([3, 1.5, 'two', None, True])
````

````python
print_table([[1, 2, 3], [4, 5, 6]])
````

````python
print_table(array("d", [1.0, 2.5, 1 / 3]))
````

````python
print_table(RECORDS)
````

````python
print_table(np.arange(5) ** 2)
````

````python
print_table(ARRAY)
````

````python
print_table(np.arange(24).reshape(2, 3, 4))  # more than 2 dimensions: one table per slice
````

````python
print_table(SERIES)
````

````python
print_table(DF, title="Cities")
````

Long and wide data shows its head and tail; columns that don't fit the width are dropped into the gap.

````python
print_table(pd.DataFrame(np.random.default_rng(1).normal(size=(1000, 40))))
````

````python
print_table(range(1000), max_rows=6)
````

Plain `print` turns numpy arrays, DataFrames and Series into tables too.

````python
rs.print("[bold]Cities[/bold]", DF)
````

## Sequence inputs

`print_list` takes lists, tuples, arrays, numpy arrays and pandas Series, and prints them as one sequence. Every input once, in the default theme.

````python
print_list([123, 1444.555, 'Test text', 4345.666])
````

````python
print_list(array("d", [1.0, 2.5, 1 / 3]))
````

````python
print_list(np.arange(5) ** 2)
````

````python
print_list(SERIES)
````

Long sequences wrap at the width; `wrap=False` keeps them on one line.

````python
print_list(SEQUENCE[:40])
````

````python
print_list(SEQUENCE[:40], wrap=False)
````

`head_tail_only` shows only the first and last `head_tail_size` elements (100 by default) around a highlighted gap.

````python
print_list(np.arange(100_000), head_tail_only=True)
````

````python
print_list(SEQUENCE, head_tail_only=True, head_tail_size=5)
````
