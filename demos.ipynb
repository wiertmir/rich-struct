# rich-struct demos

Grouped by kind of struct: under each heading the same value is printed once per theme, dark themes first, so the themes can be compared side by side. Colours don't show in GitHub's static preview.

```python
from dataclasses import dataclass, field

from array import array

import numpy as np
import pandas as pd
from rich.table import Table

import rich_struct as rs
from rich_struct import print_md, print_json, print_xml, print_object, print_yaml, print_syntax, print_table
from rich_struct import ThemeName

MARKDOWN = """Here is my top recommendation, followed by a few alternatives based on your interests:

---

### 🌟 Top Recommendation: Build a Custom Discord Bot
If you use Discord with friends or clubs, this is hands-down the most rewarding project for a high schooler.

* **Why it’s great:** You build something tangible that your friends can actually interact with. 
* **Languages to use:** Python (`discord.py`) or JavaScript/Node.js (`discord.js`).
* **What you will learn:**
  * How to read API documentation.
  * Asynchronous programming (`async`/`await`).
  * How to keep sensitive data safe using environment variables (`.env`).
  * How to host code in the cloud so it runs 24/7 (e.g., on Render or Railway).
* **Step-by-step roadmap:**
  1. **Level 1:** Make it respond to simple text commands (e.g., `!ping` replies "Pong!").
  2. **Level 2:** Add a feature like a dice roller, a coin flipper, or a mini text-trivia game.
  3. **Level 3 (Level Up):** Connect it to an external API. Make it post a random meme from Reddit, fetch the current weather, or pull game stats from an API.

---
"""

JSON = {"name": "rich-struct", "version": 0.1, "tags": ["print", "theme"], "stable": False, "license": None}

YAML = """# deployment
service:
  name: api
  replicas: 3
  debug: false
  ports: [80, 443]
  image: "ghcr.io/acme/api:1.2"
"""

XML = """<?xml version="1.0"?>
<!-- catalogue -->
<catalog updated="2026-10-11">
  <book id="bk101" lang="en">
    <title>Printing structs</title>
    <price>44.95</price>
  </book>
</catalog>
"""


@dataclass
class User:
    name: str
    age: int
    active: bool = True
    email: str | None = None
    tags: list[str] = field(default_factory=list)


OBJECT = {"user": User("Ada", 36, tags=["admin", "dev"]), "url": "https://example.com/a?b=1", "pi": 3.14159}

TABLE = Table("theme", "kind", title="A rich renderable")
TABLE.add_row("Tokyo Night", "dark")
TABLE.add_row("Light+", "light")

RECORDS = [
    {"id": 1, "name": "Ada", "score": 91.5, "admin": True},
    {"id": 2, "name": "Linus", "score": 78.25, "admin": False},
    {"id": 3, "name": "Grace", "score": None, "admin": False},
]

ARRAY = np.random.default_rng(0).normal(size=(4, 5))

DF = pd.DataFrame(
    {
        "city": ["Kraków", "Oslo", "Lima"],
        "population": [804_237, 709_037, 10_092_000],
        "area_km2": [326.8, 454.0, 2672.3],
        "capital": [False, True, True],
        "twinned": pd.to_datetime(["1991-05-18", "2003-09-01", None]),
    }
).set_index("city")

SERIES = DF["area_km2"]

```

## Table inputs

`print_table` takes lists, arrays, numpy arrays, pandas DataFrames and Series. Every input shape once, in the default theme; the per-theme comparisons follow.

```python
print_table([3, 1.5, 'two', None, True])
```

```python
print_table([[1, 2, 3], [4, 5, 6]])
```

```python
print_table(array("d", [1.0, 2.5, 1 / 3]))
```

```python
print_table(RECORDS)
```

```python
print_table(np.arange(5) ** 2)
```

```python
print_table(ARRAY)
```

```python
print_table(np.arange(24).reshape(2, 3, 4))  # more than 2 dimensions: one table per slice
```

```python
print_table(SERIES)
```

```python
print_table(DF, title="Cities")
```

Long and wide data shows its head and tail; columns that don't fit the width are dropped into the gap.

```python
print_table(pd.DataFrame(np.random.default_rng(1).normal(size=(1000, 40))))
```

```python
print_table(range(1000), max_rows=6)
```

Plain `print` turns numpy arrays, DataFrames and Series into tables too.
